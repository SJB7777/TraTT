# P0-(1) 방정식 두 줄 음미하기

음향 구동 **없이**, Takagi–Taupin 방정식 자체만 본다.
목표: $\alpha$ 와 $\eta$ 가 물리적으로 무엇인지 손에 잡히게 하는 것.

이 노트북에서 확인할 것

1. $\alpha$ = 결정을 지나는 동안 에너지가 투과빔 ↔ 회절빔 사이를 몇 번 왕복하는가
2. $\alpha=1$ 이 실제로 몇 µm 인가 (Si(111), 8 keV)
3. $\eta$ = Bragg 조건에서 벗어난 정도, 그 폭은 무엇이 정하는가
4. Laue 와 Bragg 가 왜 서로 **다른 수학 문제**인가
5. Bragg 침투 깊이 $\zeta_{\rm sat}=\min(1,1/\alpha)$ — 이 연구의 첫 정량적 예측


## 0. 방정식은 어디서 오는가

Maxwell 방정식 + 결정 안의 주기적 유전율 $\chi(\mathbf r)$ 에서 출발해서,
Bragg 조건 근처에서는 입사파 $D_0$ 와 회절파 $D_h$ 두 개만 크게 살아남는다고
두고(two-beam approximation), 포락선이 파장 규모에서 천천히 변한다고 하면
1차 연립 상미분방정식이 남는다. 이것이 Takagi–Taupin 방정식.

실단위 형태 ($z$ = 깊이):

$$
\begin{aligned}
\frac{dD_0}{dz} &= \frac{i\pi}{\lambda}\Big[\chi_0 D_0 + C\,\chi_{\bar h}\,e^{+i\phi} D_h\Big]\\[4pt]
\frac{dD_h}{dz} &= \frac{i\pi}{\lambda}\Big[C\,\chi_h\,e^{-i\phi} D_0 + (\chi_0 + 2\alpha_{\rm dev}) D_h\Big]
\end{aligned}
$$

여기서

$$\phi(z,t) = \mathbf h\cdot\mathbf u(z,t)$$

**격자 변위가 하는 일은 오직 위상 인자 $e^{\pm i\phi}$ 하나뿐이다.**
음향파는 원자를 옮길 뿐 산란능 $\chi_h$ 자체를 바꾸지 않기 때문.

이 사실이 문제 전체의 구조를 정한다 — 음향 구동은 *결합상수*가 아니라
*위상*으로 들어온다.


## 0-1. 무차원화

$\chi_0$ 는 두 식에 **똑같은 계수로** 들어오므로 소거할 수 있다 (gauge out).
정확한 진술과 유도는 바로 다음 절에. 흡수까지 넣으면 이야기가 조금 달라진다.

두께로 깊이를 정규화하고 $\zeta = z/L \in[0,1]$, 남는 무차원량은 둘뿐:

$$
\alpha = \frac{L}{\Lambda},\qquad
\Lambda = \frac{\lambda\cos\theta_B}{C\,|\chi_h|}
\qquad\text{(결정이 동역학적으로 얼마나 두꺼운가)}
$$

$$
\eta \;=\; (\text{이탈각}) \times L
\qquad\text{(Bragg 조건에서 얼마나 벗어났나)}
$$

$$
\boxed{
\begin{aligned}
\frac{dD_0}{d\zeta} &= i\pi\alpha\, e^{+i\phi}\, D_h\\[4pt]
\frac{dD_h}{d\zeta} &= i\pi\alpha\, e^{-i\phi}\, D_0 \;+\; i\eta\, D_h
\end{aligned}}
$$

이 노트북에서는 $\phi=0$. 즉 $\alpha$ 하나와 $\eta$ 하나만 본다.


## 0-1a. $\chi_0$ 소거를 한 줄씩

대칭 Laue, $\gamma_0=\gamma_h=\gamma=\cos\theta_B$ 에서 출발 식은

$$
\gamma\frac{dD_0}{dz} = \frac{i\pi}{\lambda}\Big[\chi_0 D_0 + C\chi_{\bar h}e^{+i\phi}D_h\Big]
$$
$$
\gamma\frac{dD_h}{dz} = \frac{i\pi}{\lambda}\Big[C\chi_h e^{-i\phi}D_0 + (\chi_0 + 2\beta_{\rm dev})D_h\Big]
$$

핵심은 $\chi_0$ 가 **두 식에 똑같은 계수로** 붙어 있다는 것.

### 치환

$$
k \equiv \frac{i\pi\chi_0}{\lambda\gamma},\qquad
D_0 = e^{kz}d_0,\qquad D_h = e^{kz}d_h
$$

**같은** $e^{kz}$ 를 두 빔 모두에 쓴다. 첫 식에 넣으면

$$
\gamma\big[k e^{kz}d_0 + e^{kz}d_0'\big]
= \frac{i\pi}{\lambda}\Big[\chi_0 e^{kz}d_0 + C\chi_{\bar h}e^{i\phi}e^{kz}d_h\Big]
$$

$e^{kz}$ 로 나누면

$$
\underbrace{\gamma k}_{=\;i\pi\chi_0/\lambda}\,d_0 + \gamma d_0'
= \underbrace{\frac{i\pi\chi_0}{\lambda}d_0}_{\text{같은 항}}
+ \frac{i\pi}{\lambda}C\chi_{\bar h}e^{i\phi}d_h
$$

양변에서 상쇄되어

$$
\gamma\,d_0' = \frac{i\pi}{\lambda}C\chi_{\bar h}e^{i\phi}d_h
$$

둘째 식도 같은 방식으로 $\gamma k\, d_h$ 가 $\dfrac{i\pi\chi_0}{\lambda}d_h$ 를 지운다:

$$
\gamma\,d_h' = \frac{i\pi}{\lambda}\Big[C\chi_h e^{-i\phi}d_0 + 2\beta_{\rm dev}d_h\Big]
$$

**$\chi_0$ 가 사라졌다. 근사는 없다.**

여기서 대칭 기하가 왜 필요한지가 보인다. $\gamma_0\neq\gamma_h$ 이면 $\chi_0$ 의 계수가
$\dfrac{i\pi\chi_0}{\lambda\gamma_0}$ 와 $\dfrac{i\pi\chi_0}{\lambda\gamma_h}$ 로 **서로 다르다**.
지수 인자 하나로 둘 다 지울 수 없다.

### 흡수가 있으면 — 정확한 진술

$\chi_0$ 가 복소수면 $e^{kz}$ 는 위상이 아니다:

$$
\chi_0 = \chi_0' + i\chi_0''
\quad\Rightarrow\quad
e^{kz} = \underbrace{e^{i\pi\chi_0' z/\lambda\gamma}}_{|\cdot|=1,\ \text{위상}}
\cdot
\underbrace{e^{-\pi\chi_0'' z/\lambda\gamma}}_{\text{감쇠}}
$$

따라서

$$
|D_h|^2 = |e^{kz}|^2\,|d_h|^2 = e^{-\mu z}\,|d_h|^2
$$

| $\chi_0$ 의 부분 | $\lvert D_h\rvert^2$ 에서 | 정규화 관측량에서 |
|---|---|---|
| 실수부 $\chi_0'$ | 사라짐 | 사라짐 |
| 허수부 $\chi_0''$ (흡수) | **남음** ($e^{-\mu z}$) | 사라짐 (두 빔 공통) |

우리가 쓰는 관측량 — 채널 비 $p_n$, rocking curve 모양, 거기서 뽑는 formation
time — 은 전부 정규화된 양이므로 $e^{-\mu z}$ 가 분자·분모에서 상쇄된다. 무해하다.
다만 "$|D_h|^2$ 에서 사라진다"는 표현은 부정확하다. (그리고 이 모델은 $\chi_0$ 를
아예 $0$ 으로 두므로 절대 강도도 영향받지 않는다.)

### 물리적으로: $\chi_0$ 는 어디로 갔나

$\chi_0$ 는 결정의 **평균 굴절률**이다. 두 빔이 같은 평균 매질을 지나므로 똑같은
위상을 얻는다. 간섭에 중요한 것은 두 빔의 *차이*뿐이므로 공통분은 제거할 수 있다.

**단, 완전히 사라진 것은 아니다.** $\chi_0$ 는 Bragg 각을 이동시킨다 (굴절 보정):

$$
\Delta\theta_{\rm refr} \simeq \frac{|\chi_0'|}{\sin 2\theta_B}
$$

Si 8 keV 에서 $\chi_0'\approx-1.5\times10^{-5}$, $\sin2\theta_B=0.479$ 이므로
$\Delta\theta\approx3.1\times10^{-5}\ \mathrm{rad} \approx 6.5''$. 무시할 크기가 아니다.

어디로 갔는가 — **$\eta$ 의 정의 안으로 흡수되었다.** "$\eta=0$ 이 정확히 Bragg"
라고 할 때의 그 Bragg 는 이미 굴절 보정된 각도다. 실험 데이터와 맞출 때
이 offset 을 잊으면 안 된다.


## 0-2. 무엇을 가정하고 있는가

$\chi_0$ 는 "$\chi_{\bar h}$ 와 같다"고 가정한 것이 아니라 **소거**한 것이다.
대칭 기하 $\gamma_0=\gamma_h=\cos\theta_B$ 에서 $\chi_0$ 항은 두 식에 동일한
계수로 들어가므로

$$
D_0 = e^{i\pi\chi_0 z/(\lambda\gamma_0)}\,d_0,\qquad
D_h = e^{i\pi\chi_0 z/(\lambda\gamma_0)}\,d_h
$$

로 치환하면 $d_0,d_h$ 의 방정식에서 $\chi_0$ 가 **정확히** 사라진다. 근사가 아니다.
$\chi_0$ 가 복소수(흡수)여도 공통 인자 $e^{-\mu z}$ 가 두 빔에 똑같이 곱해지므로
rocking curve 의 *모양*과 채널 비 $p_n$ 은 영향받지 않는다. 우리 관측량은 전부
정규화된 양이므로 무해하다.

단, 비대칭 기하 $\gamma_0\neq\gamma_h$ 에서는 계수가 달라 이 소거가 성립하지 않는다.
**대칭 기하 가정이 여기에 숨어 있다.**

### 진짜 숨은 가정: $\chi_{\bar h} = \chi_h^{*}$

결합 상수는 원래 $\chi_h$ 하나가 아니다:

$$
\Lambda^{-1} \;\propto\; C\sqrt{\chi_h\,\chi_{\bar h}}
$$

전자밀도가 실수(흡수 없음)이면 $\chi_{\bar h}=\chi_h^{*}$ 이므로
$\chi_h\chi_{\bar h}=|\chi_h|^2$ 가 되어 $\alpha$ 는 **실수**가 된다. 이것이 정확히
`core.py` 의 `alpha: float` 이 뜻하는 바다. 흡수가 있으면
$\sqrt{\chi_h\chi_{\bar h}}$ 는 복소수이고 $\alpha$ 도 복소수여야 한다.

Si 는 중심대칭이라 구조인자 쪽 문제는 없다. 남는 것은 흡수뿐이다.

### 이 가정이 버리는 물리

| 버려진 것 | 해로운가 |
|---|---|
| 균일 흡수 $e^{-\mu z}$ | 무해 — 공통 인자, 정규화하면 사라짐 |
| **Borrmann (anomalous transmission)** | **주의** — $\mathrm{Im}\,\chi_h$ 가 두 Bloch 모드에 서로 다른 흡수를 주는 효과 |

Borrmann 이 잠재적으로 중요한 이유: 두 모드 중 하나만 살아남으면
**모드 선택적 감쇠 = 추가 시간척도**가 생길 수 있다. 지금은 명시적으로 제외하고,
P8 의 mechanism 후보로 남겨 둔다.

### 전체 가정 목록

1. two-beam (나머지 반사 무시)
2. **대칭 기하** $\gamma_0=\gamma_h$  ← $\chi_0$ 소거가 여기에 의존
3. 평면파, 단색
4. **무흡수** $\mathrm{Im}\,\chi=0$ → $\alpha$ 실수  ← Borrmann 제외
5. $\chi_h\chi_{\bar h}=|\chi_h|^2$ (4의 따름)
6. 편광 인자 $C$ 고정 (σ: $C=1$, π: $C=\cos2\theta_B$), 편광 혼합 없음
7. 음향파는 **위상으로만** 진입 — $\chi_h$ 자체는 변하지 않음
   (탄성광학 효과, 진동에 의한 Debye–Waller 감소 무시)

7 번에서 주의할 점: 변형(strain)은 제외된 것이 **아니다**.
$\partial\phi/\partial\zeta = \mathbf h\cdot\partial\mathbf u/\partial\zeta$ 가 곧
국소 격자간격 변화이므로 $\phi$ 안에 이미 들어 있다.


```python
import matplotlib.pyplot as plt
import numpy as np

from dftt.core import required_nz, rocking, solve_bragg, solve_laue
```

## 1. $\alpha$ = 에너지가 왕복한 횟수

무변조($\phi=0$), 정확히 Bragg($\eta=0$)이면 방정식이 닫힌 형태로 풀린다:

$$
\frac{dD_0}{d\zeta} = i\pi\alpha D_h,\qquad
\frac{dD_h}{d\zeta} = i\pi\alpha D_0
$$

$D_0(0)=1,\ D_h(0)=0$ 을 넣으면

$$
D_0(\zeta) = \cos(\pi\alpha\zeta),\qquad
D_h(\zeta) = i\sin(\pi\alpha\zeta)
$$

$$
\boxed{\;|D_h(1)|^2 = \sin^2(\pi\alpha)\;}
$$

에너지가 투과빔과 회절빔 사이를 **왕복**한다. 이것이 Pendellösung.
$\alpha$ 는 결정을 지나는 동안 그 왕복이 몇 번 일어나는지를 센다.


```python
alpha = np.linspace(0, 3, 400)
I = np.array([rocking(a, 0.0, geometry="laue", n_z=1024) for a in alpha])

plt.figure(figsize=(6, 3.5))
plt.plot(alpha, I, lw=2, label="solver")
plt.plot(alpha, np.sin(np.pi * alpha) ** 2, "k--", lw=1, label=r"$\sin^2(\pi\alpha)$")
plt.xlabel(r"$\alpha = L/\Lambda$")
plt.ylabel(r"$|D_h(L)|^2$")
plt.title("Laue, on Bragg: Pendellosung")
plt.legend(frameon=False)
plt.tight_layout()
```

$\alpha=0.5$ → 에너지 전부 회절빔으로.
$\alpha=1$ → 한 바퀴 돌아 전부 투과빔으로 되돌아옴 (회절빔 $0$).

깊이에 따라 실제로 어떻게 오가는지 보자. 두께 $\zeta L$ 인 결정은
$\alpha' = \zeta\alpha$ 이므로, $\alpha$ 를 스케일해서 그리면 된다.


```python
zeta = np.linspace(0, 1, 200)

plt.figure(figsize=(6, 3.5))
for a in (0.5, 1.0, 2.0):
    Ih = [np.abs(solve_laue(a * z, 0.0, n_z=512)[1]) ** 2 for z in zeta]
    plt.plot(zeta, Ih, lw=1.6, label=rf"$\alpha={a}$")
plt.xlabel(r"$\zeta = z/L$")
plt.ylabel(r"$|D_h(z)|^2$")
plt.title("energy round trip vs depth")
plt.legend(frameon=False)
plt.tight_layout()
```

## 2. $\alpha$ 를 실단위로: Si(111), 8 keV

$$
\Lambda = \frac{\lambda\cos\theta_B}{C\,|\chi_h|},\qquad
\chi_h = -\frac{r_e\lambda^2 F_h}{\pi V},\qquad
F_{111} = 4\sqrt{2}\,f
$$

$\sin\theta_B = \lambda/2d_{111}$, Si 의 $f\simeq10.5$
($\sin\theta/\lambda \simeq 0.16\ \text{Å}^{-1}$ 에서).

> **주의.** $f$ 는 표에서 읽은 근사값이다. 발표용 최종 숫자는
> `xraylib` / XOP 로 다시 뽑을 것.


```python
r_e = 2.818e-5    # Angstrom
E_keV = 8.0
lam = 12.3984 / E_keV     # Angstrom
d111 = 3.1356             # Angstrom
V = 160.2                 # Angstrom^3
f_Si = 10.5               # form factor at this sin(theta)/lambda
C = 1.0                   # sigma polarization

theta_B = np.arcsin(lam / (2 * d111))
F = 4 * np.sqrt(2) * f_Si
chi_h = r_e * lam**2 * F / (np.pi * V)
Lambda_A = lam * np.cos(theta_B) / (C * chi_h)

print(f"lambda   = {lam:.4f} A")
print(f"theta_B  = {np.degrees(theta_B):.2f} deg")
print(f"|chi_h|  = {chi_h:.3e}")
print(f"Lambda   = {Lambda_A * 1e-4:.1f} um      <- alpha = 1 인 두께")
print()
for L_um in (1, 10, 50, 100):
    print(f"  L = {L_um:5.0f} um  ->  alpha = {L_um * 1e4 / Lambda_A:.2f}")
```

## 3. $\eta$ = Bragg 조건에서 벗어난 정도

$\eta$ 는 이탈각에 두께 $L$ 을 곱한 무차원량이다. 그래서 두꺼운 결정일수록
같은 $\eta$ 가 더 작은 실제 각도에 대응한다 — 두꺼울수록 각도적으로
예민하다는 잘 알려진 사실과 같은 이야기.

rocking curve 가 $\alpha$ 에 따라 어떻게 변하는지 보자.


```python
eta = np.linspace(-40, 40, 1201)

plt.figure(figsize=(6, 3.5))
for a in (0.1, 0.5, 1.5):
    plt.plot(eta, rocking(a, eta, geometry="laue", n_z=1024), lw=1.6, label=rf"$\alpha={a}$")
plt.xlabel(r"$\eta$")
plt.ylabel(r"$|D_h(L)|^2$")
plt.title("Laue rocking curve")
plt.legend(frameon=False)
plt.tight_layout()
```

중심 봉우리의 **폭**은 $\alpha$ 가 바뀌어도 거의 그대로다
($\eta$ 단위로 $\sim2\pi$). 이건 유한 두께가 만드는 sinc 폭이지
dynamical 폭이 아니다:

$$
D_h(1)\;\propto\;\int_0^1 e^{i\eta(1-\zeta)}\,d\zeta
\;=\;e^{i\eta/2}\,\mathrm{sinc}\!\left(\frac{\eta}{2}\right)
$$

대신 $\alpha$ 가 바뀌면 **높이**가 $\sin^2(\pi\alpha)$ 를 따라 오르내린다.

> 이건 나중에 중요해진다 — 위성 채널의 폭 $\sigma_n(t)$ 를 관측량으로 쓸 때,
> 그 폭이 무엇을 의미하는지 알고 있어야 하기 때문.


## 4. Bragg 는 완전히 다른 수학 문제

같은 방정식인데 $D_h$ 가 $-z$ 로 진행한다. 그러면 $d/ds_h = -d/d\zeta$ 이고

$$
-\frac{dD_h}{d\zeta} = i\pi\alpha\, e^{-i\phi} D_0 + i\eta D_h
$$

그리고 결정적으로 **경계조건이 양 끝에 있다**:

$$
D_0(0)=1 \quad\text{(앞면)},\qquad D_h(1)=0 \quad\text{(뒷면으로 들어오는 파 없음)}
$$

|  | Laue (투과) | Bragg (반사) |
|---|---|---|
| $D_h$ 진행 | $+z$ | $-z$ |
| 조건 위치 | 둘 다 $\zeta=0$ | 양 끝 |
| 문제 종류 | 초기값 (IVP) | **경계값 (BVP)** |

경계값 문제는 그냥 적분할 수 없다. 보통 shooting (앞면에서 $D_h(0)$ 을 추측해
뒷면이 $0$ 이 되는 값을 찾음) 을 쓰는데, 지수적으로 발산해서 잘 수렴하지 않는다.

### Riccati 로 우회

비율 하나만 추적한다:

$$
R \equiv \frac{D_h}{D_0},\qquad
\frac{dR}{d\zeta} = \frac{D_0 D_h' - D_h D_0'}{D_0^2}
$$

대입하면

$$
\boxed{\;\frac{dR}{d\zeta}
= -\,i\pi\alpha\, e^{-i\phi}
\;-\; i\eta R
\;-\; i\pi\alpha\, e^{+i\phi} R^2\;}
$$

조건은 $R(1)=0$ 하나뿐 → **다시 초기값 문제**. 뒤에서 앞으로 적분한다.
복소 ODE 한 개. 측정량은 $|R(0)|^2$.

왜 안정한가: 뒤→앞 방향이 이 Riccati 의 끌개(attracting) 방향이다.
총반사 영역에서 $|R|\to1$ 로 붙고 발산하지 않는다. shooting 의 발산 문제가
구조적으로 사라진다.

### 결과: Darwin plateau

무변조 두꺼운 결정에서 $dR/d\zeta=0$ 로 두면

$$
\pi\alpha R^2 + \eta R + \pi\alpha = 0
\;\Rightarrow\;
R = \frac{-\eta \pm \sqrt{\eta^2 - 4\pi^2\alpha^2}}{2\pi\alpha}
$$

$|\eta| < 2\pi\alpha$ 이면 판별식이 음수 → $|R|=1$, **총반사**.


```python
plt.figure(figsize=(6, 3.5))
e = np.linspace(-60, 60, 1601)
for a in (0.5, 1.0, 4.0):
    plt.plot(e, rocking(a, e, geometry="bragg", n_z=2048), lw=1.6, label=rf"$\alpha={a}$")
    for s in (-1, 1):
        plt.axvline(s * 2 * np.pi * a, color="0.7", lw=0.8, ls=":")
plt.xlabel(r"$\eta$")
plt.ylabel(r"$|R(0)|^2$")
plt.title(r"Bragg (dotted: $\pm 2\pi\alpha$)")
plt.legend(frameon=False)
plt.tight_layout()
```

## 5. 이 차이가 우리 연구에 왜 중요한가

- **Laue**: 출사 진폭이 $0$ 부터 $L$ 까지 **전 깊이의 적분**이다. 결정 전체가 기여한다.
- **Bragg**: 총반사 영역에서 파는 표면 $\sim\Lambda$ 안에서 되돌아 나온다.
  깊은 쪽은 애초에 빛을 보지 못한다.

음향 wavefront 는 앞면에서 출발해 $v_s$ 로 내려간다. 그러면

$$
\text{Laue:}\quad \text{front 가 뒷면까지 다 가야 회절 상태 확정}
\;\Rightarrow\; t_f \sim t_{tr}
$$

$$
\text{Bragg:}\quad \text{front 가 } \Lambda \text{ 만 지나면 이미 끝}
\;\Rightarrow\; t_f \sim \frac{t_{tr}}{\alpha}
$$

이것이 sweep 을 돌리기 **전에** 이미 반증 가능한 예측이다.
아래에서 "빛이 실제로 얼마나 깊이 들어가는가"를 직접 확인한다.


```python
alpha_full = 4.0
frac = np.linspace(0.02, 1.0, 60)
R_eta0 = [np.abs(solve_bragg(alpha_full * x, 0.0, n_z=2048)) ** 2 for x in frac]

plt.figure(figsize=(6, 3.5))
plt.plot(frac, R_eta0, lw=2)
plt.axvline(1 / alpha_full, color="r", ls="--", lw=1,
            label=r"$\zeta = 1/\alpha\ \ (=\Lambda/L)$")
plt.xlabel(r"slab fraction $\zeta$")
plt.ylabel(r"$|R|^2$ at $\eta = 0$")
plt.title(rf"Bragg penetration depth ($\alpha={alpha_full}$)")
plt.legend(frameon=False)
plt.tight_layout()
```

반사율이 $\zeta\sim1/\alpha$ 근처에서 이미 포화된다면, 그 아래 결정은
회절에 기여하지 않는다는 뜻이다. 음향 front 가 거기까지만 가면 formation 은 끝난다.

숫자로 확인하자.


```python
print(" alpha   zeta(90%)   zeta(99%)   1/alpha")
for A in (0.5, 1, 2, 4, 8, 16):
    x = np.linspace(1e-3, 1, 120)
    nz = required_nz(A, A)
    R = np.array([abs(solve_bragg(A * t, 0.0, n_z=nz)) ** 2 for t in x])
    z90 = x[np.argmax(R >= 0.90 * R[-1])]
    z99 = x[np.argmax(R >= 0.99 * R[-1])]
    print(f"{A:6.1f}   {z90:8.3f}   {z99:9.3f}   {1 / A:8.3f}")
```

$\alpha\ge4$ 에서 $\zeta(99\%)$ 가 $1/\alpha$ 의 $0.96\sim1.02$ 배로 맞는다.
$\alpha<1$ 이면 결정이 extinction length 보다 얇아 전체가 기여 → $1$ 로 포화.

$$
\boxed{\;\zeta_{\rm sat} = \min\!\left(1,\ \frac1\alpha\right)\;}
$$

---

### 주의 — 이것은 아직 예측일 뿐이다

위 결과는 **무변조·정적** 침투 깊이다. 실제 formation 은 변조된
시간의존 문제이므로 그대로 성립한다는 보장이 없다.

그래서 이것은 *예측*이고, P4 의 $t_f/t_{tr}$ map 이 이를 확인하거나 반증한다.
sweep 이전에 정량적 기준선을 손에 쥐었다는 것이 요점이다.

$$
\text{예측:}\qquad \frac{t_f}{t_{tr}} = \min\!\left(1,\ \frac1\alpha\right)
\quad(\text{Bragg}),
\qquad \frac{t_f}{t_{tr}} \approx 1 \quad(\text{Laue})
$$
